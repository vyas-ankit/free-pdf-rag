```python
def extract_text_only_smart(pdf_path: str, output_path: str = None) -> str:
    """
    Remove repeated text (headers/footers appear on every page)
    """
    doc = fitz.open(pdf_path)
    all_text = []
    
    for page in doc:
        blocks = page.get_text("blocks")
        page_text = [block[4].strip() for block in blocks if block[6] == 0]
        all_text.append(page_text)
    
    # Find text that appears on every page (likely header/footer)
    if all_text:
        common_lines = set(all_text[0])
        for page_text in all_text[1:]:
            common_lines &= set(page_text)
    
    # Filter out common lines
    result_lines = []
    for page_text in all_text:
        page_filtered = [line for line in page_text if line not in common_lines]
        result_lines.extend(page_filtered)
    
    result = "\n\n".join([t for t in result_lines if t])
    
    if output_path:
        with open(output_path, "w", encoding="utf-8") as f:
            f.write(result)
        print(f"✓ Saved to {output_path}")
    
    return result


extract_text_only_smart(pdf_path=pdf_path, output_path="Silver_Report_Smart.txt")
```

```python
import fitz

def extract_text_until_divider(pdf_path: str, output_path: str = None) -> str:
    """
    Scan from END backwards to find first bold divider line
    Extract only text BEFORE that line
    """
    doc = fitz.open(pdf_path)
    divider_page = None
    divider_y = None
    
    # Scan from last page backwards
    for page_num in range(len(doc) - 1, -1, -1):
        page = doc[page_num]
        page_width = page.rect.width
        
        # Get all lines on this page
        objects = page.get_drawings()
        
        for obj in objects:
            # obj is a dict, access keys not attributes
            rect = obj.get("rect")  # Changed from obj.rect
            
            if rect:
                line_width = rect.width
                
                # Found bold horizontal line spanning >80% of page
                if line_width > page_width * 0.80:
                    divider_page = page_num
                    divider_y = rect.y0
                    break
        
        # If found divider, stop searching
        if divider_page is not None:
            break
    
    # Extract text only up to the divider
    text = []
    for page_num in range(len(doc)):
        page = doc[page_num]
        
        # Skip pages after divider
        if divider_page is not None and page_num > divider_page:
            break
        
        blocks = page.get_text("blocks")
        
        for block in blocks:
            if block[6] == 0:  # Text block
                block_y = block[1]
                
                # If on divider page, stop before the line
                if divider_page == page_num and divider_y and block_y > divider_y:
                    break
                
                text.append(block[4].strip())
    
    result = "\n\n".join([t for t in text if t])
    
    if output_path:
        with open(output_path, "w", encoding="utf-8") as f:
            f.write(result)
        print(f"✓ Saved to {output_path}")
        if divider_page:
            print(f"✓ Divider found at page {divider_page + 1}")
    
    return result

# Usage
extract_text_until_divider(pdf_path=pdf_path, output_path="Silver_Report_Final.txt")
```

```python
import fitz

def extract_text_until_divider_robust(pdf_path: str, output_path: str = None) -> str:
    """
    Scan from END backwards
    Find first bold horizontal line (divider)
    Extract text BEFORE it
    """
    doc = fitz.open(pdf_path)
    divider_page = None
    divider_y = None
    
    # Scan from last page backwards
    for page_num in range(len(doc) - 1, -1, -1):
        page = doc[page_num]
        page_width = page.rect.width
        page_height = page.rect.height
        
        # Method 1: Check for drawing objects (lines)
        objects = page.get_drawings()
        
        for obj in objects:
            rect = obj.get("rect")
            if rect:
                line_width = rect.width
                line_height = rect.height
                
                # Horizontal line: wide but thin
                # AND at least 80% of page width
                if (line_width > page_width * 0.80 and 
                    line_height < 5):  # Thin line
                    
                    divider_page = page_num
                    divider_y = rect.y0
                    print(f"Found divider at page {page_num + 1}, y={divider_y}")
                    break
        
        if divider_page is not None:
            break
    
    # Extract text up to divider
    text = []
    
    for page_num in range(len(doc)):
        page = doc[page_num]
        
        # Skip pages after divider
        if divider_page is not None and page_num > divider_page:
            break
        
        blocks = page.get_text("blocks")
        
        for block in blocks:
            if block[6] == 0:  # Text block only
                block_y = block[1]
                block_text = block[4].strip()
                
                # On divider page: stop before line
                if divider_page == page_num and divider_y:
                    if block_y > divider_y:
                        break
                
                if block_text:
                    text.append(block_text)
    
    result = "\n\n".join(text)
    
    if output_path:
        with open(output_path, "w", encoding="utf-8") as f:
            f.write(result)
        print(f"✓ Saved to {output_path}")
    
    return result

# Usage
extract_text_until_divider_robust(pdf_path=pdf_path, output_path="Silver_Report_Final.txt")
```

```python
def clean_text_remove_footer(input_file: str, output_file: str = None):
    """
    Remove everything after footer markers
    """
    with open(input_file, "r", encoding="utf-8") as f:
        text = f.read()
    
    # Footer markers that always appear
    footer_markers = ["Share\n\nPrev", "Share", "Prev"]
    
    # Find first footer marker and cut there
    for marker in footer_markers:
        if marker in text:
            text = text.split(marker)[0]
            break
    
    # Clean up extra whitespace at end
    text = text.strip()
    
    if output_file:
        with open(output_file, "w", encoding="utf-8") as f:
            f.write(text)
        print(f"✓ Cleaned and saved to {output_file}")
    
    return text

# Usage
clean_text_remove_footer("Silver_Report_Final.txt", "Silver_Report_Clean.txt")
```

```python
def extract_and_clean_text(pdf_path: str, output_file: str = None):
    """
    Complete cleanup:
    1. Skip header/footer regions
    2. Stop at divider line
    3. Remove footer markers
    """
    import fitz
    
    doc = fitz.open(pdf_path)
    text = []
    
    for page_num in range(len(doc)):
        page = doc[page_num]
        page_height = page.rect.height
        
        # Skip header (top 8%) and footer (bottom 8%) regions
        skip_top = page_height * 0.03
        skip_bottom = page_height * 0.95
        
        blocks = page.get_text("blocks")
        
        for block in blocks:
            if block[6] == 0:  # Text block only
                block_y = block[1]
                block_text = block[4].strip()
                
                # Skip header/footer regions
                if not (skip_top < block_y < skip_bottom):
                    continue
                
                if block_text:
                    text.append(block_text)
    
    # Join all text
    result = "\n\n".join(text)
    
    # Remove footer markers (navigation elements)
    footer_markers = ["Share\n\nPrev", "Share", "Prev", "Next"]
    for marker in footer_markers:
        if marker in result:
            result = result.split(marker)[0]
            break
    
    # Clean extra whitespace
    result = result.strip()
    
    # Remove extra blank lines
    result = "\n\n".join([line.strip() for line in result.split("\n\n") if line.strip()])
    
    if output_file:
        with open(output_file, "w", encoding="utf-8") as f:
            f.write(result)
        print(f"✓ Extracted and cleaned: {output_file}")
    
    return result

# Usage - Single function does everything
extract_and_clean_text(pdf_path=pdf_path, output_file="Silver_Report_Clean_text_only.txt")
```

```python
import requests
import base64
import fitz
import time
from src.core.llm import get_llm_api_key

def extract_and_clean_text_with_images(pdf_path: str, output_file: str = None, max_retries: int = 3):
    """
    Complete cleanup with images:
    1. Skip header/footer regions (top 3%, bottom 5%)
    2. Stop at divider line / footer markers
    3. Remove footer markers
    4. Extract and describe images
    5. Combine text and images in reading order
    """
    
    api_key = get_llm_api_key("google").split('#')[0].strip()
    if not api_key:
        print("ERROR: No valid Google API key found")
        return None
    
    model = "gemini-flash-latest"
    url = f"https://generativelanguage.googleapis.com/v1beta/models/{model}:generateContent"
    
    doc = fitz.open(pdf_path)
    all_items = []
    
    print(f"Processing {len(doc)} pages...")
    
    # Extract text from all pages
    text_count = 0
    for page_num in range(len(doc)):
        page = doc[page_num]
        page_height = page.rect.height
        
        # Skip header (top 3%) and footer (bottom 5%) regions
        skip_top = page_height * 0.03
        skip_bottom = page_height * 0.95
        
        blocks = page.get_text("blocks")
        
        for block in blocks:
            if block[6] == 0:  # Text block only
                block_y = block[1]
                block_text = block[4].strip()
                
                # Skip header/footer regions
                if not (skip_top < block_y < skip_bottom):
                    continue
                
                if block_text:
                    all_items.append({
                        "type": "text",
                        "content": block_text,
                        "y": block_y,
                        "page": page_num
                    })
                    text_count += 1
    
    print(f"✓ Extracted {text_count} text blocks")
    
    # Remove footer markers (navigation elements)
    full_text = "\n\n".join([item["content"] for item in all_items if item["type"] == "text"])
    footer_markers = ["Share\n\nPrev", "Share", "Prev", "Next"]
    cut_index = len(full_text)
    
    for marker in footer_markers:
        if marker in full_text:
            cut_index = full_text.find(marker)
            break
    
    # Filter out text after footer marker
    all_items = [item for item in all_items if item["type"] != "text" or 
                 (full_text.find(item["content"]) + len(item["content"]) <= cut_index)]
    
    print("Processing images...")
    
    # Extract images with vision API
    image_count = 0
    for page_num in range(len(doc)):
        page = doc[page_num]
        
        for img_index in page.get_images(full=True):
            image_count += 1
            xref = img_index[0]
            pix = fitz.Pixmap(doc, xref)
            
            try:
                img_rect = page.get_image_bbox(img_index)
                img_y = img_rect.y0
            except:
                img_y = float('inf')
            
            # Convert to PNG
            if pix.n - pix.alpha < 4:
                img_data = pix.tobytes("png")
            else:
                rgb_pix = fitz.Pixmap(fitz.csRGB, pix)
                img_data = rgb_pix.tobytes("png")
            
            img_base64 = base64.standard_b64encode(img_data).decode()
            print(f"Page {page_num+1}, Image {image_count}: {len(img_base64)} bytes", end="")
            
            description = None
            for attempt in range(max_retries):
                try:
                    payload = {
                        "contents": [
                            {
                                "parts": [
                                    {"text": "Describe this image in 1-2 sentences:"},
                                    {
                                        "inline_data": {
                                            "mime_type": "image/png",
                                            "data": img_base64
                                        }
                                    }
                                ]
                            }
                        ]
                    }
                    
                    response = requests.post(
                        url,
                        headers={"X-goog-api-key": api_key},
                        json=payload,
                        timeout=30
                    )
                    
                    if response.status_code == 200:
                        result = response.json()
                        description = result["candidates"][0]["content"]["parts"][0]["text"]
                        print(f" ✓")
                        break
                    elif response.status_code == 429:
                        wait_time = (2 ** attempt) * 5
                        if attempt < max_retries - 1:
                            print(f" (waiting {wait_time}s)", end="")
                            time.sleep(wait_time)
                        else:
                            print(f" ✗")
                    else:
                        if attempt == max_retries - 1:
                            print(f" ✗")
                        if attempt < max_retries - 1:
                            time.sleep(2 ** attempt)
                
                except Exception as e:
                    if attempt == max_retries - 1:
                        print(f" ✗")
                    if attempt < max_retries - 1:
                        time.sleep(2 ** attempt)
            
            if not description:
                description = "[Image description failed]"
            
            all_items.append({
                "type": "image",
                "content": description,
                "y": img_y,
                "page": page_num
            })
            time.sleep(2)
    
    print(f"✓ Processed {image_count} images")
    
    # Sort by page and Y position (reading order)
    print("Combining text and images in reading order...")
    all_items.sort(key=lambda x: (x["page"], x["y"]))
    
    # Join all content
    result = "\n\n".join([item["content"] for item in all_items])
    
    # Clean extra whitespace and blank lines
    result = result.strip()
    result = "\n\n".join([line.strip() for line in result.split("\n\n") if line.strip()])
    
    doc.close()
    
    if output_file:
        with open(output_file, "w", encoding="utf-8") as f:
            f.write(result)
        print(f"✓ Saved to {output_file} ({len(result)} characters)")
    
    return result


# Usage:
extract_and_clean_text_with_images(pdf_path=pdf_path, output_file="Silver_Report_Complete_1.txt")
```

```python
import requests
import json
import os
from dotenv import load_dotenv
from pydantic import BaseModel

load_dotenv()

api_key = os.getenv("GOOGLE_API_KEY").split('#')[0].strip()  # Remove any comments

# Read metadata
with open("/Users/ankitvyas/free-pdf-rag/data/processed/file_3/metadata.txt", "r") as f:
    metadata_text = f.read()

# Call Google Gemini directly
url = "https://generativelanguage.googleapis.com/v1beta/models/gemini-flash-latest:generateContent"

payload = {
    "contents": [{
        "parts": [{
            "text": f"""Extract metadata EXACTLY AS-IS from this text.
Extract: title, date, theme
No changes, no inference, extract exactly as it appears.

Text:
{metadata_text}"""
        }]
    }]
}

response = requests.post(
    url,
    headers={"X-goog-api-key": api_key},
    json=payload
)

print(response.status_code)
print(response.json())
```
